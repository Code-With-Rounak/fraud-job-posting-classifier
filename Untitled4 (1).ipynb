```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score, precision_recall_curve, auc
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
from sklearn.base import clone
```

```python
fake_job = pd.read_csv("fake_job_postings.csv")
```

```python
print (fake_job.isnull().sum())
```

```python
text_cols = ['title', 'company_profile', 'description', 'requirements', 'benefits']
fake_job[text_cols] = fake_job[text_cols].fillna('')
fake_job['text'] = fake_job[text_cols].agg(' '.join, axis=1)
fake_job = fake_job.drop_duplicates(subset=text_cols + ['location']).reset_index(drop=True)
cat_cols = ['employment_type', 'required_experience',
            'required_education', 'industry', 'function']
fake_job[cat_cols] = fake_job[cat_cols].fillna('Unknown')

fake_job['country'] = fake_job['location'].fillna('Unknown').str.split(',').str[0]
cat_cols.append('country')
fake_job['has_salary'] = fake_job['salary_range'].notna().astype(int)
num_cols = ['telecommuting', 'has_company_logo', 'has_questions', 'has_salary']

X = fake_job[['text'] + cat_cols + num_cols]
y = fake_job['fraudulent']
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
pre = ColumnTransformer([
    ('tfidf', TfidfVectorizer(stop_words='english', max_features=20000,
                              ngram_range=(1, 2), min_df=3), 'text'),
    ('cat', OneHotEncoder(handle_unknown='ignore', min_frequency=10), cat_cols),
    ('num', 'passthrough', num_cols),
])
lr = Pipeline([('pre', pre),
               ('clf', LogisticRegression(max_iter=1000, class_weight='balanced'))])
lr.fit(x_train, y_train)
print(classification_report(y_test, lr.predict(x_test)))
```

```python
rn = Pipeline([('pre', clone(pre)),
               ('rlf', RandomForestClassifier(
    n_estimators=400,
    max_depth=7,
    random_state=42,
    class_weight="balanced"))])
rn.fit(x_train, y_train)
print(classification_report(y_test, rn.predict(x_test)))
```

```python
ratio = (y_train == 0).sum() / (y_train == 1).sum()
xg = Pipeline([('pre', clone(pre)),
               ('xgb', XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    scale_pos_weight=ratio,
    random_state=42,
    eval_metric="logloss"
))])
xg.fit(x_train, y_train)
print(classification_report(y_test, xg.predict(x_test)))
```

```python
from sklearn.metrics import (classification_report, precision_recall_curve, auc,
                             precision_score, recall_score, f1_score)

models = {"Logistic Regression": lr, "Random Forest": rn, "XGBoost": xg}

# Fake-probability from each model on the REAL test set
all_probs = {n: m.predict_proba(x_test)[:, 1] for n, m in models.items()}

# Soft voting = average of the 3 probabilities
avg_prob = sum(all_probs.values()) / len(all_probs)
vote_pred = (avg_prob >= 0.5).astype(int)

print("VOTING (soft)\n", classification_report(y_test, vote_pred))

# PR-AUC for every model + voting
all_probs["Voting"] = avg_prob
for name, p in all_probs.items():
    prec, rec, _ = precision_recall_curve(y_test, p)
    print(f"{name}: PR-AUC = {auc(rec, prec):.3f}")

# Comparison table for your README
rows = []
for name, m in models.items():
    p = m.predict(x_test)
    rows.append([name, precision_score(y_test, p), recall_score(y_test, p), f1_score(y_test, p)])
rows.append(["Voting", precision_score(y_test, vote_pred),
             recall_score(y_test, vote_pred), f1_score(y_test, vote_pred)])

print(pd.DataFrame(rows, columns=["Model", "Fake Precision", "Fake Recall", "Fake F1"]).round(3))
```
